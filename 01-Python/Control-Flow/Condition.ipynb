```python
# nested Condition Statements

n = int(input("Enter a number: "))

if n>0:
    print(n,"The number is positive")
    if n%2==0:
        print(n,"The number is even")
    else:
        print(n,"The number is odd")
else:
    print(n,"The number is negative or zero")
```

```python
# Example
# Determine if a year is a leap year using nested condition statement

year = int(input("Enete Year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"leap Year")
        else:
            print(year,"Not Leap Year")
    else: 
        print(year,"Leap Year")
else: 
    print(year,"Not Leap Year")
```

```python
# Simple Calculator

n = int(input("Enter First Number: "))
n1 = int(input("Enter Second Number: "))
operation = input("Enter Operation (+,-,*,/): ")    

print("First Number:", n)
print("Second Number:", n1) 
print(f"{n}{operation}{n1}")
if operation == '+':
    result = n+n1
elif operation == '-':
    result = n-n1
elif operation == '*':  
    result = n*n1   
elif operation == '/':  
    if n1!=0:
        result = n/n1
    else:
        result = "Error: Division by zero"  
else:
    result = "Invalid operation"

print("Result:", result)
```
