#  Arithmetic Operators
 
   > Addition

   > Suntraction

   > Multiplication

   > Divison

   > Floor Division

   > Modulus

   > EXponentiation


```python
a = 10 
b = 5 

add  = a+b
sub  = a-b
mul = a*b
div = a/b
floor = a//b
modulus = a%b
exponent = a**b 

print("Addition: ",add) 
print("Subtraction: ",sub)
print("Multiplication:",mul)
print("Divison: ",div)
print("Floor: ",floor)
print("Modulus: ",modulus)
print("Exponent: ",exponent)
```

# Logical Operator

 > AND

 > NOt

 > OR

```python
# AND

X = True
Y = True

result =  X and Y
print(result)

A = True
B = False

result1 =  A and B
print(result1)
```

```python
# OR
x = False
y = False

result =  x or y
print(result)

a = True
b = False

result2 =  a or b
print(result2)
```

```python
# Not

c = True
print(not c)

d = False
print(not d)
```
