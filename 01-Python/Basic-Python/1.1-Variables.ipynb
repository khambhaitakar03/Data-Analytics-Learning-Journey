##  Variable
    -> Variables are fundamental in programming used to store data that can be refrenced and manipulated in a program. In python,
       Variables are created When you assign value to them, and they do not need explicit declaration to reserve memory space.
       The declaration happen automatically when you assign a value to a vribale. 


```python
## Declaring and Assigning variables

age = 23
height = 5.9
name = "John"
is_student = True       

print("Age:", age)
print("Height:", height)
print("Name:", name)           
print("Is Student:", is_student)
```

## Naming Conventions
   > variable names should be descriptive.

   > They must start with a letter or an'_' and contains letter, number and underscore.
   
   > variable names case sensitive.
   

```python
# valid variable names
first_name = "Alice"
last_name = "Smith"

print("First Name:", first_name)    
print("Last Name:", last_name)  
```

```python
# invalid variable names
# 2age = 30
# first-name = "Bob"
# @name = "Charlie"
```

```python
# case sensitivity
name = "David"
Name = "Eve"

print("name:", name)    
print("Name:", Name)
```

```python
# Under Standing Variable Types
# python is dynamically typed, type of variable is determined at runtime.

age = 25
heght = 5.9
name = "John"
is_student = True

print(type(name))
```

```python
# type checkig and conversion
age = 25
print(type(age)) 

# Type conversion
age_str = str(age)
print(age_str)
print(type(age_str))  
```

```python
# dynamic typing

var = 10
print(var,type(var))

var = "Hello"   
print(var,type(var))

var = 3.14  
print(var,type(var)) 
```

```python
# input
age = input("Enter your age: ")
print(age,type(age)) 

a = int(input("Enter a number: "))
print(a,type(a))
```

```python
# Simple Calculator

n1 = float(input("Enter first number: "))
n2 = float(input("Enter second number: "))

sum = n1 + n2
difference = n1 - n2
product = n1 * n2
quotient = n1 / n2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
