## Syntax and semantics in pythjon

## 1.Syntax :
        syntax is refers to the set of rules that defines the combination of symbol that one considered to be correctly structured.
        In simple terms,syntax is about the correct arragment of words and symbol ina code

## 2.Semantics 
        Semantics refers to the the meaning or the interpretation of the symbol,characters and command in language.
        it is about what the code is supposed to do whent  it runs.



```python
#Basic syntax Rules in Python
#Case sensitive : Python is Case sensitive 

name="Harsh"
Name="Khambhaitakar"

print(name)
print(Name)
```

## 3.Indentation
    -> Indentation in python is used to define the structure and hierarchy of the code. Unlike many other  Programming languages that
       use braces {} to delimit blocks of code, python uses indentation to determine the grouping of statements within block must be 
       indented at the same level.
    -> Python use indentation to define blocks of code. Consistent use of spaces (comnly 4) or tab is required.

```python
age = 20

if age >= 18:
    print(age)  # space called Indentation
```

```python
# line continuation : use \ to continue the line

total = 1 + 2 + 3 +\
        4+5+6
print(total)
```

```python
# multiple statements in single line : use ; to separate the statements
x=5; y=10; z=x+y    

print(z)
```

```python
# understand semantics of Python : 
# Python is a dynamically typed language, which means that you don't have to declare the type of a variable when you create it. 
# The interpreter infers the type based on the value assigned to the variable.

age=32
type(age)
```

```python
# type inference

var1 = 10
print(type(var1)) 
var1 = "Hello"
print(type(var1))
```
