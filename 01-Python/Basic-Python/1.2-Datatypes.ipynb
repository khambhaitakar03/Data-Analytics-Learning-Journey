# Data Types

1. Defination

   -> Data Types are classification of data which tell the compiler or interpreter how the programmer
      intends to use the data.

   -> They determine the type of operation theat can be performed on the data, the values that the
      data can take, and the data can take,and the amount of memory needed to store the data.

2. Importance of data type in programming
   > Explanation : 
   
      -> data types ensure the data is stored in an efficient way.

      -> They help in performing correct operation on data.
      
      -> proper use of data types can prevent errors and bugs in the program.
